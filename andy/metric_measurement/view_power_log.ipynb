```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("dark_background")

power_df = pd.read_csv('../../power_log_ME005.csv')
power_df.head(5)

x = power_df['Time (s)'].values
gpu_power = power_df['GPU Power (mW)']
cpu_power = power_df['CPU Power (mW)']
average_power = power_df['Average Power (mW)']

plt.plot(x, gpu_power, label='gpu')
plt.plot(x, cpu_power, label='cpu')
plt.plot(x, average_power, label='average')
plt.ylim(0, 500)
plt.legend()
plt.show()
```
